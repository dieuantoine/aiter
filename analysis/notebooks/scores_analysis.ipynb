```python
import pandas as pd

df = pd.read_csv("../../data/results_hyp_3.csv")
df
```

```python
df.groupby("model_id")["score"].describe().sort_values("mean")
```

```python
df.groupby("request")["score"].describe().sort_values("mean")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.histplot(df['score'], bins=20, kde=True)
plt.title("Histogramme global des scores")
plt.xlabel("Score")
plt.ylabel("Fréquence")
plt.grid(True)
plt.show()
```

```python
df['nb_mots'] = df['response'].str.split().str.len()
correlation = df['nb_mots'].corr(df['score'])
print(f"Corrélation (nombre de mots vs score) : {correlation:.3f}")
```
